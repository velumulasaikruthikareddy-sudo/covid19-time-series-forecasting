```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load dataset
df = pd.read_csv("../dataset/owid-covid-data.csv")

df.head()
```

```python
india_data = df[df['location'] == 'India']

# Convert date column
india_data['date'] = pd.to_datetime(india_data['date'])

# Show first 5 rows
india_data.head()
```

```python
plt.figure(figsize=(12,6))

plt.plot(india_data['date'], india_data['total_cases'])

plt.title("Total COVID Cases in India")
plt.xlabel("Date")
plt.ylabel("Total Cases")

plt.show()
```

```python
from sklearn.linear_model import LinearRegression

# Remove missing values
india_data = india_data.dropna(subset=['total_cases'])

# Create numeric days
india_data['days'] = np.arange(len(india_data))

# Features and target
X = india_data[['days']]
y = india_data['total_cases']

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict next 30 days
future_days = np.arange(len(india_data), len(india_data) + 30).reshape(-1, 1)

predictions = model.predict(future_days)

# Plot graph
plt.figure(figsize=(12,6))

plt.plot(india_data['days'], y, label='Actual Cases')
plt.plot(future_days, predictions, label='Predicted Cases')

plt.title("COVID-19 Future Prediction")
plt.xlabel("Days")
plt.ylabel("Total Cases")

plt.legend()

plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import math

# Predictions on existing data
train_predictions = model.predict(X)

# Calculate errors
mae = mean_absolute_error(y, train_predictions)
mse = mean_squared_error(y, train_predictions)
rmse = math.sqrt(mse)

print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
```
